## Base

```python
T = 5000
theta_min = 0.5
gamma = 100 / T  # ≈ 0.001
alpha_val = 0.9
tau = 12

num_layers = 12
r_base = 8
r_max = 2 * r_base  # 16
delta = r_max / (num_layers - 1)  # 16 / 11 ≈ 1.455

r_list = [min(int(delta * (l - 1)), r_max) for l in range(1, num_layers + 1)]
```

```python
def get_layer_drop_prob(r, step, theta_min=0.5, tau=12):
    """
    Calculate the drop probability for a given layer and step.
    """
    alpha = 0.9 if r >= tau else 1.0
    prob = alpha * ((1 - theta_min) * math.exp(-gamma * step) + theta_min)

    return prob
```

```python
t = 1
prev_n = n_tokens = 197

for i in range(num_layers):
    r = r_list[i]
    new_n = prev_n - r
    print(f" --- Layer {i}: --- ")
    print(f"prev_n: {prev_n}, new_n: {new_n}, reduced: {r}")
    
    # alpha(l)
    if r >= tau:
        alpha_l = alpha_val
        print("### Tau met: using alpha ###")
    else:
        alpha_l = 1.0

    # theta_{t,l}
    current_theta = alpha_l * ((1 - theta_min) * math.exp(-gamma * t) + theta_min)

    print(f"Current theta: {current_theta:.6f}")
    
    prev_n = new_n
    # t += 1 
```
